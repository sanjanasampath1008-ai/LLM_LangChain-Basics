## Building LLM Applications using LangChain Prompt Templates and Chains

### Objective
- Understand LangChain fundamentals.
- Create reusable prompt templates.
- Build LLM chains.
- Integrate LangChain with Ollama.
- Generate and evaluate chatbot responses.

```python
import os
import sys
import pandas as pd

from IPython.display import display

sys.path.insert(
    0,
    os.path.abspath("..")
)

from langchain_chatbot import (
    llm,
    PROMPT_TEMPLATES,
    build_chain,
    summarization_chain,
    question_answering_chain
)

print("Libraries imported successfully!")
```

```python
for name, template in PROMPT_TEMPLATES.items():

    print("\nPROMPT TEMPLATE:", name.upper())

    print(template)
```

```python
main_chain = build_chain("general")

query = "What is Artificial Intelligence?"

response = main_chain.invoke({

    "user_input": query,
    "college_name": "ABC College"

})

print("User Query:", query)

print("\nChatbot Response:")

print(response)
```

```python
text = """
Artificial Intelligence is a branch of computer science.
It enables machines to perform tasks that usually require
human intelligence. AI is used in education, healthcare,
transportation and many other industries.
"""

summary = summarization_chain.invoke({

    "text": text

})

print("Original Text:")
print(text)

print("\nSummarized Text:")
print(summary)
```

```python
context = """
LangChain is a framework used to develop applications
powered by Large Language Models. It provides prompt
templates, chains and integrations with different models.
"""

question = "What is LangChain?"

answer = question_answering_chain.invoke({

    "context": context,
    "question": question

})

print("Question:", question)

print("\nAnswer:", answer)
```
